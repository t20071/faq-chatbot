# 04 · Persistent SQLite index

Stores the LLM Zoomcamp FAQ in `database/faq.db` with `sqlitesearch`, so it survives kernel restarts.
The same thing from the command line: `uv run build-faq-db`.

```python
from llm_zoomcamp_dtc.ingest import load_faq_data

documents = load_faq_data()
print(f"Loaded {len(documents)} documents")
```

```python
docs_llm = [doc for doc in documents if doc["course"] == "llm-zoomcamp"]
print(f"LLM Zoomcamp: {len(docs_llm)} documents")
```

```python
from llm_zoomcamp_dtc.sqlite_index import DATABASE_PATH, open_sqlite_index

sqlite_index = open_sqlite_index()  # database/faq.db
```

```python
import time

for doc in docs_llm:
    sqlite_index.add(doc)
    print(f'Added: {doc["question"][:60]}...')
    time.sleep(0.5)

sqlite_index.close()
print(f"Done. Index saved to {DATABASE_PATH}")
```
